# Exploratory analysis of the synthetic MFS data
Shows that fraud and legitimate behaviour **overlap** (no single feature separates them).

```python
import pandas as pd
df = pd.read_csv('data/processed/generated_transactions_features.csv')
print(len(df), 'transactions;', f'{df.true_fraud.mean():.2%} true fraud;', f'{df.is_fraud.mean():.2%} labelled fraud (label noise)')
```

```python
df.groupby('true_fraud')[['amount_vs_sender_avg','sender_velocity','receiver_velocity','receiver_fan_in_24h','new_relationship','rapid_forwarding','device_changed','receiver_age_days']].mean().round(2)
```

```python
f = df[df.true_fraud==1]; l = df[df.true_fraud==0]
print('stealth fraud (amount <1.8x normal):', f'{(f.amount_vs_sender_avg<1.8).mean():.0%}')
print('legit look-alikes (amount >2.5x normal):', f'{(l.amount_vs_sender_avg>2.5).mean():.1%}')
print('fraud on a new device:', f'{f.device_changed.mean():.0%}', '| legit on a new device:', f'{l.device_changed.mean():.1%}')
```
